1. Load and Display

```python
import pandas as pd
df = pd.read_csv('./data/house_price.csv')
df.head(2)
```

2. Show Statistical Summary

```python
df.describe()
```

3. Filter Data

```python
df_filtered = df[df['bedrooms'] > 2]
df_filtered
```

4. Sort Data


```python
df_sort = pd.DataFrame(df)
df_sorted = df_sort.sort_values(by='sales_price')
df_sorted
```

5. Calculate Average Sales Price

```python
df['sales_price'].mean()
```

6. Add a New Column


```python
df['price_per_sqr_meter']= df['sales_price'] / df['sq_meter']
df
```

7. Group Data


```python
grouped_df = df.groupby('neighborhood')['sales_price'].mean()
grouped_df.reset_index()
```

8. Visualize Data

```python
import matplotlib.pyplot as plt
grouped_df.plot(kind='bar', x='neighborhood', y='sales_price')
plt.title('Average sales price per neighorhood')
plt.xlabel('Neighborhood')
plt.ylabel('Average sales price')
plt.show()
```

9. Find Maximum Values


```python
max_house_price = df[df['sales_price'] == df['sales_price'].max()]
max_house_price
```

10. Export Data

```python
df.to_csv('./data/new_data.csv', index='false')
```
